```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('NBA_team_payroll_2026.csv')
```

```python
plt.figure(figsize=(20, 6))
x = df.Team
plt.figure(figsize=(20, 6))
x = df.Team
y = df['Total Cash']
plt.xlabel('Team', fontsize=18)
plt.ylabel('Total Cash', fontsize=18)
plt.title('NBA Team Payroll by Team (2026)', fontsize=20)
plt.bar(x, y)
plt.tight_layout()
plt.show()
plt.xlabel('Team', fontsize=18)
plt.ylabel('Total Cash', fontsize=18)
plt.title('NBA Team Payroll by Team (2026)', fontsize=20)
plt.bar(x, y)
plt.tight_layout()
plt.show()
```
