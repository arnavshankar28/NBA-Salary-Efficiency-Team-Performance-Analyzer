```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('NBA_team_stats.csv')
```

```python
plt.figure(figsize=(80, 20))
x = df['Team']
y = df['PTS']
plt.xlabel('Team', fontsize=18)
plt.ylabel('PPG', fontsize=18)
plt.title('NBA PPG per Team (2026)', fontsize=20)
plt.bar(x, y)
plt.tight_layout()
plt.show()
```
